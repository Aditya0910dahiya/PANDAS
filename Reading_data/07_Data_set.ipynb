```python
data={"name":["raam","shyam","ghanshyam","yugh","aditya","akash","shiva","utkarsh","simran","navjot"],
      "age":[10,20,30,77,90,45,44,99,100,65],
      "salary":[10000,200000,430000,530000,670000,130000,1500000,220000,33300020,12345600],
      "performance score":[99,97,87,85,67,88,49,91,88,56]

}
data_frame=pd.DataFrame(data)

high_salary=data_frame[data_frame["salary"]>50000]
print("employ with salary>50000")
print(high_salary)

# filtering rows salary>50k & age >30

filtered = data_frame[
    (data_frame["age"] > 30) &
    (data_frame["salary"] > 50000)
]

print("employ list age >30+salary>50000")
print(filtered)

# using or condition 
filtered_or=data_frame[(data_frame["age"]>35) | (data_frame["performance score"]>90)]
print("employees older than 35 or performance score>90")
print(filtered_or)
```
