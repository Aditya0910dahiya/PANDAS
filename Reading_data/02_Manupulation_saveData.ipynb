```python
import pandas as pd
```

# how to manupulate and how to save data in different files

```python
data={"name":["raam","shyam","ghanshyam"],
      "age":[10,20,30],
      "city":["nagpur","mumbai","delhi"]
}
data_frame=pd.DataFrame(data)
print(data_frame)

data_frame.to_csv("output.csv")
```

```python
data={"name":["raam","shyam","ghanshyam"],
      "age":[10,20,30],
      "city":["nagpur","mumbai","delhi"]
}
data_frame=pd.DataFrame(data)
print(data_frame)

data_frame.to_csv("output2.csv",index=False)
```

```python
data={"name":["raam","shyam","ghanshyam"],
      "age":[10,20,30],
      "city":["nagpur","mumbai","delhi"]
}
data_frame=pd.DataFrame(data)
print(data_frame)

data_frame.to_excel("output.xlsx",index=False)
```

```python
data={"name":["raam","shyam","ghanshyam"],
      "age":[10,20,30],
      "city":["nagpur","mumbai","delhi"]
}
data_frame=pd.DataFrame(data)
print(data_frame)

data_frame.to_json("output2.json",index=False)
```
