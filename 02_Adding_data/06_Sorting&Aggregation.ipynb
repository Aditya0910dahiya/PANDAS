# Sorting data

```python
data = {
    "name": ["raam", "shyam", "ghanshyam","navjot"],
    "age": [35, 44, 99, 65],
    "salary": [10000, 200000, 430000, 12345600],
}

data_frame = pd.DataFrame(data)
data_frame.sort_values(by="age",ascending=False,inplace=True)
print("sorted age by descending ")
print(data_frame)
```

```python
data = {
    "name": ["raam", "shyam", "ghanshyam","navjot"],
    "age": [35, 44, 99, 65],
    "salary": [10000, 200000, 430000, 12345600],
}

data_frame = pd.DataFrame(data)
data_frame.sort_values(by=["age","salary"],ascending=[True,False],inplace=True)
print("sorted age by descending ")
print(data_frame)
```
