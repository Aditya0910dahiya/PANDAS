# additng new column using square brackets

```python
data={"name":["raam","shyam","ghanshyam","yugh","aditya","akash","shiva","utkarsh","simran","navjot"],
      "age":[10,20,30,77,90,45,44,99,100,65],
      "salary":[10000,200000,430000,530000,670000,130000,1500000,220000,33300020,12345600],
      "performance score":[99,97,87,85,67,88,49,91,88,56]

}
data_frame=pd.DataFrame(data)
print(data_frame)

data_frame["Bonus"]=data_frame["salary"]*0.1
print(data_frame)

# now column is added of bonus
```
