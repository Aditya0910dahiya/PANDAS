# It can be represented  by two type
# 1-NaN (not a number)
# 2- None(for object data types)


# if come True means Value is missing
# if False then value is present

```python
import pandas as pd

data = {
    "name": ["raam", "shyam", "ghanshyam", None, "aditya", "akash", "shiva", "utkarsh", "simran", "navjot"],
    "age": [10, 20, 30, None, 90, 45, 44, 99, 100, 65],
    "salary": [10000, 200000, 430000, None, 670000, 130000, 1500000, 220000, 33300020, 12345600],
    "performance_score": [99, 97, 87, None, 67, 88, 49, 91, 88, 56]
}

data_frame = pd.DataFrame(data)

print(data_frame)

print(data_frame.isnull())
```

# dropna()

```python
data = {
    "name": ["raam", "shyam", "ghanshyam", None, "aditya", "akash", "shiva", "utkarsh", "simran", "navjot"],
    "age": [10, 20, 30, None, 90, 45, 44, 99, 100, 65],
    "salary": [10000, 200000, 430000, None, 670000, 130000, 1500000, 220000, 33300020, 12345600],
    "performance_score": [99, 97, 87, None, 67, 88, 49, 91, 88, 56]
}
data_frame = pd.DataFrame(data)
print(data_frame)

data_frame.dropna(inplace=True)
print(data_frame)
# now nan values are removed
```

# fillna Method

```python
data = {
    "name": ["raam", "shyam", "ghanshyam", None, "aditya", "akash", "shiva", "utkarsh", "simran", "navjot"],
    "age": [10, 20, 30, None, 90, 45, 44, 99, 100, 65],
    "salary": [10000, 200000, 430000, None, 670000, 130000, 1500000, 220000, 33300020, 12345600],
    "performance_score": [99, 97, 87, None, 67, 88, 49, 91, 88, 56]
}

data_frame = pd.DataFrame(data)
print(data_frame)

data_frame.fillna(0,inplace=True)
print(data_frame)
```
